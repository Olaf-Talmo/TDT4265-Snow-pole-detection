```python
dataset_path_lidar = "/home/omtalmo/Olaf_TTK4265/Poles/lidar"
```

```python
from ultralytics import YOLO

# Initialize YOLOv12 model
model = YOLO('yolov12s.yaml')

# Train the model on our LiDAR dataset
results = model.train(
    data=f'{dataset_path_lidar}/labels/data.yaml',
    epochs=100,              # Number of epochs for training
    imgsz=640,               # Image size
    batch=16,                # Batch size
    workers=8,               # Number of workers for data loading
    device='0',              # GPU device
    val=True,                # Perform validation
    plots=True,              # Generate plots
    save=True,               # Save model checkpoints
    project='lidar_detection',  # Project name
    name='yolov12s_lidar',   # Run name
    exist_ok=True,           # Overwrite existing project/name
    pretrained=True,         # Use pre-trained weights
    patience=20,             # Early stopping patience
    verbose=True             # Print verbose output
)

```
