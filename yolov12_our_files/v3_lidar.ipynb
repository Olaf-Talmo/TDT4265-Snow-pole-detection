### make a new class that is calles reflector that is put inside each of the poles

```python
import os
import numpy as np
from pathlib import Path

def generate_reflector_labels(
    labels_dir: str,
    output_dir: str,
    reflector_scale: float = 0.2,
    random_offset: float = 0.05
):
    """
    Generate reflector labels inside existing pole labels
    
    Args:
        labels_dir: Directory with original YOLO labels
        output_dir: Where to save new labels
        reflector_scale: Size of reflector relative to pole
        random_offset: Max random offset from center (as fraction of pole size)
    """
    
    # Create output directory
    output_path = Path(output_dir)
    output_path.mkdir(parents=True, exist_ok=True)
    
    # Process each label file
    for label_file in Path(labels_dir).glob('*.txt'):
        new_lines = []
        
        with open(label_file, 'r') as f:
            for line in f:
                parts = line.strip().split()
                class_id = int(parts[0])
                
                if class_id != 0:  # Skip if not pole
                    continue
                    
                # Keep original pole annotation
                x, y, w, h = map(float, parts[1:])
                new_lines.append(f"0 {x} {y} {w} {h}")
                
                # Generate reflector with slight random offset
                rx_offset = np.random.uniform(-random_offset, random_offset) * w
                ry_offset = np.random.uniform(-random_offset, random_offset) * h
                
                rx = x + rx_offset
                ry = y + ry_offset
                rw = w * reflector_scale
                rh = h * reflector_scale
                
                # Add reflector annotation
                new_lines.append(f"1 {rx} {ry} {rw} {rh}")
        
        # Save new labels
        output_file = output_path / label_file.name
        with open(output_file, 'w') as f:
            f.write('\n'.join(new_lines) + '\n')

# Usage example
if __name__ == "__main__":
    labels_dir = "/home/edvarsa/pole_detection/lidar_1/train/labels"
    output_dir = "/home/edvarsa/pole_detection/lidar_1/train/new_labels"
    
    generate_reflector_labels(
        labels_dir=labels_dir,
        output_dir=output_dir,
        reflector_scale=0.2,  # Reflector is 20% of pole size
        random_offset=0.05    # Can shift up to 5% from center
    )

    
```

```python
import cv2
import matplotlib.pyplot as plt
from pathlib import Path
import os

%matplotlib inline

def visualize_poles_and_reflectors(dataset_path_lidar, num_images=5):
    """Visualize poles and reflectors in training images."""
    
    # Setup paths
    train_img_dir = Path(dataset_path_lidar) / "train" / "images"
    train_label_dir = Path(dataset_path_lidar) / "train" / "labels"
    
    # Get first num_images from training set
    image_files = sorted(list(train_img_dir.glob("*.png")))[:num_images]
    
    # Create figure
    fig, axes = plt.subplots(num_images, 1, figsize=(12, 4*num_images))
    if num_images == 1:
        axes = [axes]
    
    for idx, img_path in enumerate(image_files):
        # Load image
        img = cv2.imread(str(img_path))
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        h, w = img.shape[:2]
        
        # Load corresponding label file
        label_path = train_label_dir / f"{img_path.stem}.txt"
        
        # Draw boxes
        if label_path.exists():
            with open(label_path) as f:
                for line in f:
                    cls_id, x_center, y_center, width, height = map(float, line.strip().split())
                    
                    # Convert normalized coordinates to pixel coordinates
                    x1 = int((x_center - width/2) * w)
                    y1 = int((y_center - height/2) * h)
                    x2 = int((x_center + width/2) * w)
                    y2 = int((y_center + height/2) * h)
                    
                    # Draw box with different colors for poles (red) and reflectors (green)
                    color = (1, 0, 0) if cls_id == 0 else (0, 1, 0)
                    label = "Pole" if cls_id == 0 else "Reflector"
                    
                    # Draw rectangle and label
                    axes[idx].add_patch(plt.Rectangle((x1, y1), x2-x1, y2-y1, 
                                                    fill=False, color=color, linewidth=2))
                    axes[idx].text(x1, y1-5, label, color=color, 
                                 backgroundcolor='white', fontsize=8)
        
        axes[idx].imshow(img)
        axes[idx].set_title(f"Image {idx+1}: {img_path.name}")
        axes[idx].axis('off')
    
    plt.tight_layout()
    plt.show()

# Run visualization 
visualize_poles_and_reflectors(dataset_path_lidar, num_images=1)
```
