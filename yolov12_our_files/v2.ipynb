```python
from ultralytics import YOLO

# Load your pretrained model
model = YOLO('/home/omtalmo/Olaf_TTK4265/TDT4265-Snow-pole-detection/yolov12_our_files/yolov12/yolov12/runs/detect/train32/weights/best.pt')  

# --- optional: narrow or expand the search space -------------------------
search_space = {
    "lr0": (1e-5, 1e-1),           # base LR
    "lrf":      (0.01, 1.0), 
    "momentum": (0.7, 0.97),
    "weight_decay": (0.0, 1e-2),
    "mosaic": (0.0, 1.0),          # aug probabilities
    "cls": (0.20, 4.00),
    "dfl": (0.30, 1.00),
}

# --- launch the run -------------------------------------------------------
results = model.tune(
    data = f'{dataset_path}/data.yaml',  # your dataset (train/val splits)
    epochs=10,                # max epochs per trial
    iterations=10,           # number of evolutionary trials
    imgsz=640,
    optimizer="AdamW",
    space=search_space,
    plots=True,               # save fitness & scatter plots
    save=True,                # keep checkpoints
    val=True,                 # run full val each epoch
)
```
